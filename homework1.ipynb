Q4. Missing values
Explanation:
First, we calculate the number of missing values in every column using isna().sum().
Then, isna().any() identifies which columns contain at least one missing value. Finally, we count those columns.

```python
# Q4: Number of columns with missing values

# Count missing values in each column
missing_values = df.isna().sum()

print("Missing values in each column:")
print(missing_values)

# Count how many columns contain at least one missing value
columns_with_missing = df.isna().any().sum()

print("\nNumber of columns with missing values:", columns_with_missing)
```

Q6. Median value of horsepower
Explanation:
First, we calculate the median of the horsepower column.
Then, we find the most frequent horsepower value using mode().

Next, we use fillna() to replace missing horsepower values with this most frequent value.

Finally, we calculate the median again and compare it with the original median.



```python
# Q6: Median horsepower before and after filling missing values

# Calculate the median horsepower before filling missing values
median_before = df["horsepower"].median()

# Find the most frequent horsepower value
most_frequent_horsepower = df["horsepower"].mode()[0]

# Fill missing horsepower values with the most frequent value
df["horsepower"] = df["horsepower"].fillna(most_frequent_horsepower)

# Calculate the median horsepower after filling missing values
median_after = df["horsepower"].median()

print("Median horsepower before filling missing values:", median_before)
print("Most frequent horsepower:", most_frequent_horsepower)
print("Median horsepower after filling missing values:", median_after)

# Compare the two median values
if median_after > median_before:
    print("\nAnswer: Yes, it increased")
elif median_after < median_before:
    print("\nAnswer: Yes, it decreased")
else:
    print("\nAnswer: No, it did not change")
```

```python
# Q7: Linear regression and sum of weights

import numpy as np

# Select all cars from Asia
asia_cars = df[df["origin"] == "Asia"]

# Select vehicle_weight and model_year
# Then select the first 7 rows
X = asia_cars[["vehicle_weight", "model_year"]].head(7).to_numpy()

print("X:")
print(X)

# Calculate X transpose multiplied by X
XTX = X.T @ X

print("\nXTX:")
print(XTX)

# Calculate the inverse of XTX
XTX_inv = np.linalg.inv(XTX)

print("\nInverse of XTX:")
print(XTX_inv)

# Create the y array provided in the homework
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

# Calculate the weights:
# w = (XTX)^(-1) * X^T * y
w = XTX_inv @ X.T @ y

print("\nWeights (w):")
print(w)

# Calculate the sum of all elements in w
sum_of_weights = w.sum()

print("\nSum of weights:", sum_of_weights)
```
